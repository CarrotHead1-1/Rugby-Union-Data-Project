```python
from pyspark.sql import SparkSession

#load data from the landing schema 
df_matches_1 = spark.read.csv(
    '/Volumes/rugby_data_dev/rugby_landing/raw_data/matchResults2015-2018.csv',
    header = True, inferSchema = True
)

df_matches_2 = spark.read.csv(
    '/Volumes/rugby_data_dev/rugby_landing/raw_data/matchResults2018-2026.csv',
    header = True, inferSchema = True
)

df_match_data = spark.read.csv(
    '/Volumes/rugby_data_dev/rugby_landing/raw_data/premiershipMatchData22-26.csv',
    header = True, inferSchema = True
)

#create delta tables and ingest into the bronze schema 
df_matches_1.write.format('delta').mode('overwrite').saveAsTable('rugby_data_dev.rugby_bronze.match_results_raw_2015_2018')
df_matches_2.write.format('delta').mode('overwrite').saveAsTable('rugby_data_dev.rugby_bronze.match_results_raw_2018_2026')
df_match_data.write.format('delta').mode('overwrite').saveAsTable('rugby_data_dev.rugby_bronze.match_results_raw_2022_2026')

df_matches_1.printSchema()
df_matches_2.printSchema()
df_match_data.printSchema()
```
