```python


#load data from landing schema and ingest using schemas 
df_matches_1 = spark.read.csv(
    '/Volumes/rugby_data_dev/rugby_landing/raw_data/matchResults2015-2018.csv',
    header = True, schema = matchSchema
)

df_matches_2 = spark.read.csv(
    '/Volumes/rugby_data_dev/rugby_landing/raw_data/matchResults2018-2026.csv',
    header = True, schema = matchSchema
)

df_match_data = spark.read.csv(
    '/Volumes/rugby_data_dev/rugby_landing/raw_data/premiershipMatchData22-26.csv',
    header = True, schema = matchDataSchema
)

#add basic data quality checks and add metadata

#records by file
print(f"2015-2018: {df_matches_1.count()}")
print(f"2018-2026: {df_matches_2.count()}")
print(f"2022-2026: {df_match_data.count()}")

#check for Null values
for df_name, df in [('df_matches_1', df_matches_1), ('df_matches_2', df_matches_2), ('df_match_data', df_match_data)]:
    print(f"\n=== Null Checks for {df_name}")
    df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns]).show()

#add meta data 
df_matches_1 = df_matches_1.withColumn('ingestTimestamp', current_timestamp()).withColumn("sourceFile", lit("matchResults2015-2018"))
df_matches_2 = df_matches_2.withColumn('ingestTimestamp', current_timestamp()).withColumn("sourceFile", lit("matchResults2018-2026"))
df_match_data = df_match_data.withColumn('ingestTimestamp', current_timestamp()).withColumn("sourceFile", lit("matchResults2022-2026"))
```

```python
#create delta tables and ingest into the bronze schema 
df_matches_1.write.format('delta').mode('overwrite').option("mergeSchema", "true").saveAsTable('rugby_data_dev.rugby_bronze.match_results_raw_2015_2018')
df_matches_2.write.format('delta').mode('overwrite').option("mergeSchema", "true").saveAsTable('rugby_data_dev.rugby_bronze.match_results_raw_2018_2026')
df_match_data.write.format('delta').mode('overwrite').option("mergeSchema", "true").saveAsTable('rugby_data_dev.rugby_bronze.match_results_raw_2022_2026')

df_matches_1.printSchema()
df_matches_2.printSchema()
df_match_data.printSchema()
```
